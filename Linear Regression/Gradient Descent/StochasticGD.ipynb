```python
import numpy as np
```

```python
class StochasticGD:
    
    def __init__(self , learning_rate = 0.01 , epochs = 100):
        self.learning_rate = learning_rate
        self.epochs = epochs
        self.weights = None
        self.intercept = None
        
    def fit(self , X ,y):
        X = np.asarray(X)
        y = np.array(y)
        
        self.weights = np.zeros(X.shape[1])
        self.intercept = 0
        
        for i in range(self.epochs):
            idx = np.random.permutation(X.shape[0])
            for id in idx :
                x_idx = X[id] 
                y_idx = y[id]
                y_pred = np.dot(x_idx , self.weights) + self.intercept
                error = y_pred  - y_idx
                slope_weight = error * x_idx 
                slope_intercept = error  
                
                self.weights  -= self.learning_rate * slope_weight
                self.intercept -= self.learning_rate * slope_intercept 
                
        return self.weights , self.intercept
    
    def predict(self , X):
        return np.dot(np.asarray(X) , self.weights )  +self.intercept  
```

```python
np.random.seed(42)

X = np.arange(1, 101).reshape(-1, 1)
y = 2 * X.ravel() + 3 + np.random.randn(100)


X_mean = X.mean(axis=0)
X_std = X.std(axis=0)
X_scaled = (X - X_mean) / X_std


sgd = StochasticGD(learning_rate=0.01, epochs=100)
sgd.fit(X_scaled, y)


y_pred = sgd.predict(X_scaled)

print("Weights:", sgd.weights)
print("Intercept:", sgd.intercept)
print("First 5 predictions:", y_pred[:5])
```

```python
from sklearn.metrics import r2_score, mean_squared_error

print("R2:", r2_score(y, y_pred))
print("MSE:", mean_squared_error(y, y_pred))
```

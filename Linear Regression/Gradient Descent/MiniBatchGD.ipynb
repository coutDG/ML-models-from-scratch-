```python
import numpy as np
```

```python
class MiniBatchGD:
    """  Every epoch we updates based on a batch size"""
    def  __init__(self, learning_rate = 0.01 , epochs = 1000 , batch_size = 32):
        self.learning_rate = learning_rate
        self.epochs = epochs
        self.batch_size = batch_size
        self.weights = None
        self.intercept = None
        
    def fit(self , X , y):
        X = np.asarray(X)
        y = np.asarray(y)
        
        self.weights = np.zeros(X.shape[1])
        self.intercept = 0
        
        for i in range(self.epochs):
            idx = np.random.permutation(X.shape[0])
            X = X[idx]
            y = y[idx]
            
            for start in range(0 , X.shape[0] , self.batch_size):
                end = start + self.batch_size
                X_batch = X[start:end]
                y_batch = y[start:end]
                
                y_pred = X_batch @ self.weights + self.intercept
                error = y_pred - y_batch
                
                slope_weights = (2/len(y_batch)) * (X_batch.T @ error)
                slope_intercept = (2/len(y_batch)) * np.sum(error)
                self.weights -= self.learning_rate*slope_weights
                self.intercept -= self.learning_rate*slope_intercept
                
        return self.weights , self.intercept
    
    def predict(self , X):
        return np.asarray(X)@self.weights + self.intercept
```
